## ① 한 줄 정의

> 교과서 말 말고 **내 말로** 한 줄. 내 말로 못 쓰면 아직 모르는 것.

- **수익률**: 돈이 몇 % 늘었거나 줄었나. 오늘 ÷ 어제 − 1
- **CAGR**: 여러 해 동안 번 걸 "매년 평균 몇 %씩 불어난 셈인지"로 바꾼 것 (복리 기준)
- **변동성**: 수익률이 매일 얼마나 들쭉날쭉한가 = 수익률의 표준편차

## ② 왜 생겼나

> 이 지표가 없으면 어떤 판단을 잘못하게 되나? (예: 수익률만 보면 ○○를 놓친다)

- **수익률을 %로 보는 이유**: 100만 원에서 5만 원 번 것과 1억에서 5만 원 번 건 실력이 다름 → 금액 말고 비율로 봐야 크기 다른 돈끼리 비교 가능
- **CAGR이 필요한 이유**: 3년 돌린 전략과 10년 돌린 전략을 비교하려면 "1년 기준"으로 맞춰야 함. 그냥 전체 수익률 ÷ 년수로 나누면 복리를 빼먹어서 실력을 **부풀리게** 됨
- **변동성이 필요한 이유**: 1년에 똑같이 10% 벌어도, 꾸준히 오른 계좌 A와 +5%·−4% 오락가락한 계좌 B는 다름. B는 무서워서 중간에 팔아버릴 수 있음 → 그 "무서움"을 숫자로 만든 것

## ③ 공식

> 공식을 쓰고 **기호마다 뜻**을 옆에 적기. 면접에서 화이트보드에 쓸 수 있을 정도로.

**수익률**
$$r_t = \frac{P_t}{P_{t-1}} - 1 = \frac{P_t - P_{t-1}}{P_{t-1}}$$
- $P_t$: 오늘 금액, $P_{t-1}$: 어제 금액 (두 식은 같은 식)

**CAGR**
$$\text{CAGR} = \left(\frac{\text{마지막 금액}}{\text{처음 금액}}\right)^{1/\text{년수}} - 1$$
- 원리: 처음 × (1 + x)^년수 = 마지막 → 원금은 **한 번만**, 불어나는 비율 (1+x)만 년수만큼 곱함
- 주의: (100 × x)⁵ 처럼 원금까지 제곱하면 틀림!

**변동성**
$$\sigma_{일} = \text{STDEV}(r_1, \dots, r_n), \qquad \sigma_{연} = \sigma_{일} \times \sqrt{252}$$
- 252 = 1년 거래일 수
- 왜 √252?: 날마다 수익률이 서로 독립이면 **분산**은 일수만큼 더해짐(×252) → 표준편차는 분산의 제곱근이라 **√252** 배

## ④ 손계산

> 숫자 10개 정도로 엑셀·종이에 직접 계산하고 과정과 답을 여기 옮겨 적기.

**문제 1** 계좌 100 → 102 → 99 → 103 → 101

| 날 | 계산 | 수익률 |
|---|---|---|
| 1 | (102 − 100) ÷ 100 | **+2.00%** |
| 2 | (99 − 102) ÷ 102 = −3/102 = −0.0294 | **−2.94%** |
| 3 | (103 − 99) ÷ 99 = 4/99 = 0.0404 | **+4.04%** |
| 4 | (101 − 103) ÷ 103 = −2/103 = −0.0194 | **−1.94%** |

- 실수했던 것 ① −3/102 는 그 자체가 비율(−0.0294) → %로 쓰려면 ×100 해서 −2.94%. 분수 뒤에 %를 바로 붙이면 100배 작아짐
- 실수했던 것 ② 넷째 날 101 − 103 = **−2** (떨어졌으니 마이너스)
- 엑셀: A1~A5에 금액 → B2 `=A2/A1-1` 아래로 복사 → `=STDEV(B2:B5)` = **3.29%**
- 해석: 하루 평균 수익률 0.29%인데 하루하루 ±3% 정도씩 흔들림 → 평균보다 흔들림이 10배 넘게 큼 = 많이 출렁이는 계좌

**문제 2** 100만 원 → 5년 뒤 150만 원
- 식: 100 × (1 + x)⁵ = 150 → (1 + x)⁵ = 1.5 → 1 + x = 1.5^(1/5) → x = 1.5^(1/5) − 1 = **8.45%**
- 검산: 1.0845⁵ = 1.5002 ✓
- 50% ÷ 5 = 10% 보다 낮은 이유: **복리**. 8.45%씩만 늘어도 뒤로 갈수록 불어난 돈에 붙어서 5년 뒤 50%가 됨. 단순히 나눈 10%는 복리를 빼먹은 과대평가

## ⑤ 파이썬 재현

> ④ 를 파이썬 5줄 안팎으로 다시 계산 → 손계산과 답이 같은지 확인. 다르면 어디서 틀렸는지 찾기.

```python
import numpy as np

p = np.array([100, 102, 99, 103, 101])
r = p[1:] / p[:-1] - 1               # 날마다 수익률 = 오늘 ÷ 어제 − 1 (p[1:] 는 둘째 날부터, p[:-1] 은 마지막 날 빼고)
print("수익률(%):", np.round(r * 100, 2))         # [ 2.   -2.94  4.04 -1.94]

vol_day = r.std(ddof=1)               # ddof=1 → 엑셀 STDEV 와 같은 '표본' 표준편차
print("일 변동성:", round(vol_day, 4))            # 0.0329
print("연 변동성:", round(vol_day * np.sqrt(252), 4))  # × √252

cagr = 1.5 ** (1 / 5) - 1             # 5년에 1.5배
print("CAGR:", round(cagr, 4))                    # 0.0845
print("단순 나누기:", 0.5 / 5)                     # 0.10 → 복리 무시해서 과대평가
```

## ⑥ 프로젝트 코드 읽기

> `src/risk.py` 또는 `src/backtest.py` 에서 이 지표를 계산하는 줄을 찾아 복사하고, **한 줄씩 주석**으로 뜻 달기.

```python
import numpy as np
import pandas as pd

# src/backtest.py 의 perf_metrics() 에서 가져온 줄 — 연습용 계좌로 똑같이 돌려봄
eq = pd.Series([100, 102, 99, 103, 101], dtype=float)   # 계좌 가치 (프로젝트에선 백테스트 결과 equity)

r = eq.pct_change().dropna()          # pct_change = 오늘 ÷ 어제 − 1. 첫날은 '어제'가 없어 NaN → dropna 로 제거
yrs = len(r) / 252                    # 수익률 개수(거래일 수) ÷ 252 = 몇 년 돌렸나
cagr = (eq.iloc[-1] / eq.iloc[0]) ** (1 / yrs) - 1   # (마지막 ÷ 처음)^(1/년수) − 1 → 손계산 1.5^(1/5) − 1 과 같은 구조
vol = r.std() * np.sqrt(252)          # pandas .std() 는 기본이 표본 표준편차(엑셀 STDEV 와 같음) × √252 = 연 변동성

print(r.round(4).tolist())
print("연 변동성:", round(vol, 4))
# 참고: 4일짜리 데이터로 '연' CAGR 을 구하면 엄청 큰 숫자가 나옴 → 짧은 기간 CAGR 은 의미 없음 (⑧ 한계)
print("CAGR(4일 데이터, 의미 없음):", round(cagr, 2))
```

## ⑧ 한계

> 이 지표가 **놓치는 것**은? (리스크관리 면접은 한계를 꼭 물어봄)

- **수익률·CAGR은 중간 과정을 안 봄**: 처음과 끝만 봐서, 중간에 −50% 폭락했다가 회복해도 CAGR은 똑같음 → 그래서 MDD 가 따로 필요 (다음 주)
- **CAGR은 시작·끝 날짜에 민감**: 폭락 직후에 시작하면 좋아 보이고, 고점에 시작하면 나빠 보임. 기간이 짧으면 의미 없음 (4일 데이터로 연 CAGR 구하면 말도 안 되는 숫자)
- **변동성은 오르는 흔들림도 위험으로 침**: +5% 와 −5% 를 똑같이 취급. 투자자가 진짜 무서운 건 내려가는 쪽
- **변동성은 '평소' 흔들림만 봄**: 코로나 같은 가끔 오는 큰 폭락(꼬리 위험)은 잘 안 잡힘 → VaR·ES 가 필요한 이유
- **√252 는 날마다 독립이라는 가정**: 실제로는 하락이 며칠 연속 이어지기도 해서 연 변동성이 과소평가될 수 있음
